```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
a = np.array([2, 1])
b = np.array([1, 3])
```

```python
result = a + b
result
```

```python
origin = np.array([0, 0])
```

```python
plt.quiver(*origin, *a, color='blue', scale=1, scale_units='xy', angles='xy')
plt.quiver(*a, *b, color='red', scale=1, scale_units='xy', angles='xy')
plt.quiver(*origin, *result, color='green', scale=1, scale_units='xy', angles='xy')
plt.xlim(0, 5); plt.ylim(0, 5)
plt.grid(True)
plt.show()
```
